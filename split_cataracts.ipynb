```python
import os
import random
from utils.paths import path_data_cataracts_split, path_data_cataracts_original, path_data_cataracts
seed = 42
```

```python
videos_cataracts_seg = [f for f in os.listdir(os.path.join(path_data_cataracts_original)) if os.path.isdir(os.path.join(path_data_cataracts_original, f))]
videos_cataracts = [f for f in os.listdir(os.path.join(path_data_cataracts, "video_frames")) if os.path.isdir(os.path.join(path_data_cataracts, "video_frames", f))]
```

```python
videos_cataracts_seg = [int(f[len("Video"):]) for f in videos_cataracts_seg]
videos_cataracts = [int(f[len("train"):]) for f in videos_cataracts]
```

```python
def to_set(lst):
    s = set(lst)
    assert len(s) == len(lst)
    return s
videos_cataracts_seg = to_set(videos_cataracts_seg)
videos_cataracts = to_set(videos_cataracts)
```

```python
test_val_split = videos_cataracts_seg
train_cases = videos_cataracts - test_val_split
```

```python
random.seed(seed)
val_cases = random.sample(list(test_val_split), int(0.5 * len(test_val_split)))
test_cases = list(test_val_split - set(val_cases))
```

```python
train_cases = sorted(list(train_cases))
val_cases = sorted(val_cases)
test_cases = sorted(test_cases)
```

```python
def write_cases_to_file(cases, filename):
    os.makedirs(path_data_cataracts_split, exist_ok=True)
    with open(os.path.join(path_data_cataracts_split, filename), "w") as f:
        for case in cases:
            f.write(f"{case:02d}\n")

write_cases_to_file(train_cases, "cataracts_train.txt")
write_cases_to_file(val_cases, "cataracts_val.txt")
write_cases_to_file(test_cases, "cataracts_test.txt")
```
