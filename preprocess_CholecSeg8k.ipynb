```python
import cv2, os
import numpy as np

from PIL import Image
from tqdm import trange
from data.cholec_vis import get_cholecseg8k_colormap

from matplotlib import pyplot as plt
```

```python
INPUT_PATH = "/local/scratch/clmn1/data/cholecseg8k/"
OUT_PATH = "/local/scratch/clmn1/data/cholecseg8k_preprocessed_3/"
```

```python
color_classes = get_cholecseg8k_colormap()

palette = get_cholecseg8k_colormap()
```

```python
for patient in os.listdir(INPUT_PATH):
    print(patient)
    if not os.path.isdir(os.path.join(INPUT_PATH, patient)):
        continue
    for f in os.listdir(os.path.join(INPUT_PATH, patient)):
        first_frame = int(f[len("videoXX_"):])
        path = os.path.join(INPUT_PATH, patient, f)

        for frame in trange(first_frame, first_frame + 80):
            label_img = Image.open(os.path.join(path, f"frame_{frame}_endo_color_mask.png")).convert("RGB")
            label = np.array(label_img)

            new_label = np.zeros((label.shape[0], label.shape[1]), dtype=np.uint8)
            valid = np.zeros((label.shape[0], label.shape[1]), dtype=bool)

            for class_idx, color in enumerate(color_classes):
                mask = np.all(label == color, axis=-1)
                new_label[mask] = class_idx
                valid = valid | mask
            
            #if not np.all(valid):  # who cares xD
            #    print(f"Warning: some pixels in {patient}/{f}/frame_{frame}_endo_color_mask.png do not belong to any class!")
            #    invalid_pixels = label[~valid]
            #    colors, counts = np.unique(invalid_pixels.reshape(-1, 3), axis=0, return_counts=True)
            #    for c, cnt in zip(colors, counts):
            #        print(f"  Invalid color {tuple(c)} : {cnt} pixels")

            os.makedirs(os.path.join(OUT_PATH, patient, f), exist_ok=True)
            assert new_label.dtype == np.uint8
            assert new_label.ndim == 2
            output_mask = Image.fromarray((new_label.astype(np.uint8)))
            output_mask.putpalette(np.uint8(palette))
            output_mask.save(os.path.join(OUT_PATH, patient, f, f"frame_{frame}_mask.png"))
        
```
