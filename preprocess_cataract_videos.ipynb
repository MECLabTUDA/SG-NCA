```python
from fileinput import filename
import json
from bidict import bidict
import torch
from utils.paths import path_data_cataracts_split, path_data_cat_sg, path_data_cataracts_full_videos, path_data_cataracts_full_videos_preprocessed
import os
from PIL import Image
import torchvision.transforms as T
import imageio
import tqdm
```

```python
videos = os.listdir(os.path.join(path_data_cataracts_full_videos, "micro"))
```

```python
for video in videos:
    if not video.endswith(".mp4"):
        continue
    video_path = os.path.join(path_data_cataracts_full_videos, "micro", video)
    vid = imageio.get_reader(video_path,  'ffmpeg')
    for frame_idx in tqdm.trange(vid.count_frames(), desc=f"Processing {video}"):
        preprocessed_video_path = os.path.join(path_data_cataracts_full_videos_preprocessed, video.removesuffix(".mp4"), f"{frame_idx:08d}.png")
        if os.path.exists(preprocessed_video_path):
            continue
        try:
            frame = vid.get_data(frame_idx)
        except Exception as e:
            print(f"Error reading frame {frame_idx} from {video}: {e}")
            continue
        frame = Image.fromarray(frame).resize((256,256))
        os.makedirs(os.path.dirname(preprocessed_video_path), exist_ok=True)
        frame.save(preprocessed_video_path)
    vid.close()
```
