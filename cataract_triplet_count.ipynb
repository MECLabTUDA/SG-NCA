```python
from data.cataracts_scene_graph2 import CataractSceneGraphDataset
from tqdm import tqdm
import pandas as pd

d = CataractSceneGraphDataset("train", temporal_window=5)

num_encounters = {}
for video, frame_idx, relations in d.video_frame_tuples:
    for triple in relations:
        triple = tuple(triple)
        num_encounters[triple] = num_encounters.get(triple, 0) + 1
```

```python
df = pd.DataFrame(list(num_encounters.items()), columns=["Triplet", "Count"])
df.sort_values("Count", ascending=False, inplace=True)
df
```

```python
triplets = list(num_encounters.keys())
```

```python
firsts = set()
lasts = set()
for a, b, c in triplets:
    assert a != c
    firsts.add(a)
    lasts.add(c)
print("Subjects:", firsts)
print("Objects:", lasts)
```
