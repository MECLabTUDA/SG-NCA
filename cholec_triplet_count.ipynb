```python
from data.cholec_scene_graph import CholecSceneGraphDataset
from tqdm import tqdm
import pandas as pd

d = CholecSceneGraphDataset(split="train", temporal_window=2)

num_valid = 0
num_encounters = {}
for _, _,  sg in tqdm(d.video_frame_tuples):
    for connection in sg:
        subject, verb, object_ = connection
        num_encounters[(subject, verb, object_)] = num_encounters.get((subject, verb, object_), 0) + 1
df = []
for k, v in sorted(num_encounters.items()):
    df.append((d.get_str_triplet(k), v))
```

```python
df = pd.DataFrame(df, columns=["Triplet", "Count"])
df.sort_values("Count", ascending=False, inplace=True)
df
```
